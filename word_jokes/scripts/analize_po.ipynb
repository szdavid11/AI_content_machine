```python
import pandas as pd
```

```python
df_corpus = pd.read_csv('../hu_50k.txt', on_bad_lines='skip', dtype=str, sep=' ').dropna()
df_bad_words = pd.read_csv('../unknown.txt', on_bad_lines='skip', dtype=str, sep=' ', header=None).dropna()
df_bad_words.columns = ['szavak']
df_corpus = df_corpus[~df_corpus['szavak'].isin(df_bad_words['szavak'])]
df_corpus[['szavak']].to_csv('../hu_50k.cleaned.txt', index=False, header=False)
```

```python
df = pd.read_csv('../hu_50k.with_pos.tsv', sep='\t', header=None)
```

```python
df.head()
df.columns = ['szo', 'po']
df['po'] = df['po'].str.split(',')
df_exp = df.explode('po')
df_exp.head()
```

```python
all_po = df_exp['po'].value_counts() 
```

```python
for po in all_po.index:
    print(po)
    print(df_exp[df_exp['po'] == po].head()['szo'].to_list())
    print('-'*100)
```
