```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load the model (google/embeddinggemma-300m)
model_name = "google/embeddinggemma-300m"
model = SentenceTransformer(model_name)
print(f"Loaded model: {model_name}")
```

```python
def embedding_distance(text1: str, text2: str, metric: str = "cosine") -> float:
    """
    Compute the distance between two texts using the loaded embedding model.
    Supported metrics: 'cosine', 'euclidean'.
    """
    emb1 = model.encode([text1], convert_to_numpy=True, normalize_embeddings=False)[0]
    emb2 = model.encode([text2], convert_to_numpy=True, normalize_embeddings=False)[0]
    if metric == "cosine":
        # Normalize for cosine distance
        emb1 = emb1 / (np.linalg.norm(emb1) + 1e-12)
        emb2 = emb2 / (np.linalg.norm(emb2) + 1e-12)
        return float(1.0 - np.dot(emb1, emb2))
    elif metric == "euclidean":
        return float(np.linalg.norm(emb1 - emb2))
    else:
        raise ValueError(f"Unsupported metric: {metric}")

# Example usage:
d1 = embedding_distance("cat", "dog")
d2 = embedding_distance("cat", "table")
print(f"Cosine distance (cat, dog): {d1:.4f}")
print(f"Cosine distance (cat, table): {d2:.4f}")
```
